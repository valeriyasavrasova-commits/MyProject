```python
text = "My World. Nice World. A happy World! Indeed."
```

```python
# Define all substring's occurrences in a string
def all_positions(haystack, needle):
# the main idea is to iterate through the string to the end and to find all appearances of the substring
    pos, hits = -1, []
    while True:
        pos = haystack.find(needle, pos + 1)
        if pos == -1:
            break
        hits.append(pos)
    return hits

if __name__ == "__main__":
    print("Строка:", text)

# Contributor's note (Mikhail's):
# Note: In a Jupyter Notebook (.ipynb), __name__ is always "__main__". 
# Therefore, this guard is redundant here (can clear that), but it is kept out of habit because it is best practice for standard .py scripts
# Besides that, the idea and the way it was implemented are really nice!
```
